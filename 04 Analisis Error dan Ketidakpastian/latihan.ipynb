```python
import numpy as np
import matplotlib.pyplot as plt

def x(t):
    return t**3

def v(t):
    return 3 * t**2

def forward_difference(t, h):
    return (x(t + h) - x(t)) / h

def central_difference(t, h):
    return (x(t + h) - x(t - h)) / (2 * h)

t = 10
h = 0.1

v_excat = v(t)
v_forward = forward_difference(t, h)
v_central = central_difference(t, h)

error_forward = abs(v_forward - v_excat)
error_central = abs(v_central - v_excat)

print(f"Kecepatan eksak          = {v_excat} m/s\n")
print(f"Kecepatan forward        = {v_forward} m/s")
print(f"Absolute Error (Forward) = {error_forward}\n")

print(f"Kecepatan central        = {v_central} m/s")
print(f"Absolute Error (Central) = {error_central}")
```

```python
h_values = np.logspace(-1, -5, 5)

error_forward = []
error_central = []

for h in h_values:
    v_forward = forward_difference(t, h)
    v_central = central_difference(t, h)

    e_forward = abs(v_forward - v_excat)
    e_central = abs(v_central - v_excat)

    error_forward.append(e_forward)
    error_central.append(e_central)

plt.loglog(h_values, error_forward, 'o-', label="Forward Difference")

plt.loglog(h_values, error_central, 's-', label="Central Difference")

plt.xlabel("Step size h")
plt.ylabel("Absolute Error")

plt.title("Error Kecepatan Numerik")
plt.legend()
plt.grid(True)

plt.show()
```
