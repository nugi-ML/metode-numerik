```python
def midpoint_rule(f, a, b, N):
    h = (b - a) / N

    total = 0.0

    for i in range(N):
        x_mid = a + (i + 0.5) * h
        total += f(x_mid)
    
    return h * total


def trapezoidal_rule(f, a, b, N):
    h = (b - a) / N

    total = f(a) + f(b)

    for i in range(1, N):
        x = a + i * h
        total += 2 * f(x)
    
    return (h / 2) * total


def simpson_rule(f, a, b, N):
    if N % 2 != 0:
        raise ValueError("N harus genap untuk Simpson's Rule")
    
    h = (b - a) / N

    total = f(a) + f(b)

    for i in range(1, N):
        x = a + i * h

        if i % 2 == 0:
            total += 2 * f(x)
        
        else:
            total += 4 * f(x)
    
    return (h / 3) * total
```

### **Usaha**

Sebuah benda mengalami gaya:

$$ F(x) = 5x^3 + 2x $$

Hitung usaha dari $x=0$ sampai $x=3 \ m$

```python
# Usaha
def F(x):
    return 5*x**3 + 2*x

a = 0
b = 3
N = 10

mid  = midpoint_rule(F, a, b, N)
trap = trapezoidal_rule(F, a, b, N)
simp = simpson_rule(F, a, b, N)

print(f"Midpoint Rule    = {mid:.4f} J")
print(f"Trapezoidal Rule = {trap:.4f} J")
print(f"Simpson's Rule   = {simp:.4f} J")
```

### **Massa Batang Tidak Homogen**

Sebuah batang panjang $L=2 \ m$ mempunyai rapat massa:

$$ \rho (x) = 3 + 2x + x^2 $$

Jika luas penampang:

$$ A = 0.01 \ m^2 $$

hitung massa batang dengan integral numerik.

```python
# Massa batang tidak homogen
def density(x):
    return 3 + 2*x + x**2

def integral(x):
    return density(x) * A

a = 0
b = 2 # panjang batang (L)
N = 10

A = 0.01 # Luas penampang

mid  = midpoint_rule(integral, a, b, N)
trap = trapezoidal_rule(integral, a, b, N)
simp = simpson_rule(integral, a, b, N)

print(f"Midpoint Rule    = {mid:.4f} kg")
print(f"Trapezoidal Rule = {trap:.4f} kg")
print(f"Simpson's Rule   = {simp:.4f} kg")
```

### **Probabilitas Kuantum**

Diberikan:

$$\psi(x) = A e^{-x^2}.$$

Gunakan integral numerik untuk menentukan konstanta normalisasi $A$ sehingga:

$$\int_{-\infty}^{\infty} |\psi(x)|^2 \, dx = 1.$$

Karena komputer tidak dapat menghitung sampai infinity secara langsung, gunakan:

$$-10 \le x \le 10.$$

**Penurunan Persamaan**

Karena

$$|\psi(x)|^2 = A^2 e^{-2x^2},$$

maka

$$A^2 \int_{-10}^{10} e^{-2x^2} \, dx = 1.$$

Jadi:

$$A = \frac{1}{\sqrt{\int_{-10}^{10} e^{-2x^2} \, dx}}.$$

```python
# Probabilitas Kuantum
import numpy as np

def probabilitas(x):
    return np.exp(-2 * x**2)

a = -10
b = 10
N = 1000

integral = simpson_rule(probabilitas, a, b, N)

A = 1 / np.sqrt(integral)

normalisasi = A**2 * integral # Hasil harus 1

print(f"Hasil integral          = {integral:.4f}")
print(f"Konstanta Normalisasi A = {A:.4f}")
print(f"Hasil Normalisasi       = {normalisasi:.4f}")
```

### **Energi Potensial**

Sebuah partikel bergerak dalam potensial:

$$ V(x) = \frac{1}{2} k x^2 $$

Hitung integral energi potensial rata-rata:

$$ \langle V \rangle = \int V(x) |\psi(x)|^2 \ dx $$

Gunakan fungsi gelombang Gaussian dan integral numerik.

**Penurunan Persamaan**

Fungsi gelombang Gaussian:

$$ V(x) = \frac{1}{2} k x^2 $$

dan fungsi gelombang Gaussian:

$$\psi(x) = A e^{-x^2}$$

Dari soal sebelumnya, fungsi gelombang harus ternormalisasi:

$$\int_{-\infty}^{\infty} |\psi(x)|^2 \, dx = 1$$

Kita telah memperoleh:

$$\boxed{A = \left(\frac{2}{\pi}\right)^{1/4}}$$

sehingga

$$|\psi(x)|^2 = A^2 e^{-2x^2} = \sqrt{\frac{2}{\pi}} e^{-2x^2}$$

Nilai ekspektasi energi potensial adalah

$$\boxed{\langle V \rangle = \int_{-\infty}^{\infty} V(x) |\psi(x)|^2 \, dx}$$

Substitusikan

$$V(x) = \frac{1}{2} k x^2$$

dan

$$|\psi(x)|^2 = \sqrt{\frac{2}{\pi}} e^{-2x^2}.$$

Maka:

$$\langle V \rangle = \int_{-\infty}^{\infty} \frac{1}{2} k x^2 \sqrt{\frac{2}{\pi}} e^{-2x^2} \, dx.$$

Karena komputer tidak dapat mengintegralkan sampai tak hingga secara langsung, gunakan:

$$-10 \le x \le 10.$$

Jadi secara numerik:

$$\boxed{\langle V \rangle \approx \int_{-10}^{10} \frac{1}{2} k x^2 \sqrt{\frac{2}{\pi}} e^{-2x^2} \, dx}$$

```python
# Energi Potensial

def psi(x):
    return A * np.exp(-x**2)

def V(x):
    return 0.5 * k * x**2

def integral(x):
    return  V(x) * abs(psi(x))**2

A = (2 / np.pi)**(1/4)
k = 10

a = -10
b = 10

N = 1000

V_avg = simpson_rule(integral, a, b, N)

print(f"Konstanta Normalisasi A    = {A}")
print(f"Konstanta pegas k          = {k}")
print(f"Energi potensial rata-rata = {V_avg}")
```
