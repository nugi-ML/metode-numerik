# **Metode Bisection**

Bisection merupakan salah satu metode pencarian akar yang paling sederhana dan aman.

Metode ini memanfaatkan **Teorema Nilai Antara**

Jika fungsi kontinu dan:

$$ f(a)f(b) < 0 $$

maka terdapat setidaknya satu akar pada interval

$$ [a,b] $$

Misalkan:

$$ f(a) < 0 $$

dan 

$$ f(b) > 0 $$

Kita mengambil titik tengah:

$$ c = \frac{a + b}{2} $$

Kemudian mengevaluasi:

$$ f(c) $$

Jika

$$ f(a)f(c) < 0 $$

akar berada pada:

$$ [a,c] $$

Jika tidak, akar berada pada:

$$ [c,b] $$

Proses tersebut diulangi.

### **Implementasi Bisection dengan Python**

Kita gunakan:

$$ f(x) = x^3 - 2x - 5 $$

```python
def f(x):
    return x**3 - 2*x - 5

def bisection(f, a, b, tol=1e-10, max_iter=100):
    if f(a) * f(b) > 0:
        raise ValueError("Interval tidak mengurung akar")
    
    for i in range(max_iter):
        c = (a + b) / 2

        if abs(f(c)) < tol:
            return c, i + 1
        
        if f(a) * f(c) < 0:
            b = c
        else:
            a = c
    
    return c, max_iter

root, iterations = bisection(f, 1, 3)

print(f"Akar    = {root}")
print(f"Iterasi = {iterations}")
```

# **Metode Regula Falsi**

Bisection menggunakan titik tengah:

$$ c = \frac{a + b}{2} $$

Regulasi Falsi mencoba menggunakan informasi kemiringan fungsi.

Titik baru diperoleh dari perpotongan garis lurus yang menghubungkan:

$$ (a,f(a)) $$

dan 

$$ (b,f(b)) $$

dengan sumbu $x$.

Rumusnya:

$$ x_r = b - \frac{f(b)(b - a)}{f(b) - f(a)} $$

atau bentuk ekuivale:

$$ x_r = \frac{af(b) - bf(a)}{f(b) - f(a)} $$

```python
def regula_falsi(f, a, b, tol=1e-10, max_iter=100):
    if f(a) * f(b) > 0:
        raise ValueError("Interval tidak mengurung akar")

    for i in range(max_iter):
        fa = f(a)
        fb = f(b)

        c = (a * fb - b * fa) / (fb - fa)
        fc = f(c)

        if abs(fc) < tol:
            return c, i + 1

        if fa * fb < 0:
            b = c
        else:
            a = c

    return c, max_iter

root, iterations = regula_falsi(f, 1, 3)

print(f"Akar    = {root}")
print(f"iterasi = {iterations}")
```

# **Metode Newton-Raphson**

Newton-Raphson meggunakan informasi turunan.

Persamaannya:

$$ \boxed{x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}} $$

Metode ini dapat diturunkan dari **Deret Taylor**.

Kita mulai dari:

$$ f(x_{n+1}) $$

Dengan ekspansi Taylor:

$$ f(x_{n+1}) \approx f(x_n) + f'(x_n)(x_{n+1} - x_n) $$

Karena kita ingin:

$$ f(x_{n+1}) = 0 $$

maka:

$$ 0 \approx f(x_n) + f'(x_n)(x_{n+1} - x_n) $$

Sehingga:

$$ f'(x_n)(x_{n+1} - x_n) = -f(x_n) $$

dan:

$$ \boxed{x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}} $$

### **Implementasi Newton-Raphson**

Misalkan:

$$ f(x) = x^3 - 2x - 5 $$

Turunannya:

$$ f'(x) = 3x^2 - 2 $$

Python:

```python
def f(x):
    return x**3 - 2*x - 5

def df(x):
    return 3*x**2 - 2

def newton_raphson(f, df, x0, tol=1e-10, max_iter=100):
    x = x0

    for i in range(max_iter):
        fx = f(x)
        dfx = df(x)

        if dfx == 0:
            raise ZeroDivisionError("Turunan bernilai nol")

        x_new = x - fx / dfx

        if abs(x_new - x) < tol:
            return x_new, i + 1

        x = x_new

    return x, max_iter

root, iterations = newton_raphson(f, df, 2)

print(f"Akar    = {root}")
print(f"iterasi = {iterations}")
```

# **Metode Secant**

Secant dapat dianggap sebagai versi Newton-Raphson yang **tidak membutuhkan turunan analitik.**

Secant mengganti turunan dengan pendekaan:

$$ f'(x_n) \approx \frac{f(x_n) - f(x_{n-1})}{x_n - x_{n-1}} $$

Sehingga:

$$ \boxed{x_{n+1} = x_n - f(x_n) \frac{x_n - x_{n-1}}{f(x_n) - f(x_{n-1})}} $$

```python
def secant(f, x0, x1, tol=1e-10, max_iter=100):
    for i in range(max_iter):
        f0 = f(x0)
        f1 = f(x1)

        if f1 - f0 == 0:
            raise ZeroDivisionError("Pembagi bernilai nol")

        x2 = x1 - f1 * (x1 - x0) / (f1 - f0)

        if abs(x2 - x1) < tol:
            return x2, i + 1

        x0 = x1
        x1 = x2

    return x1, max_iter

root, iterations = secant(f, 1, 3)

print(f"Akar    = {root}")
print(f"iterasi = {iterations}")
```
