# **1. Pendahuluan**

Dalam matematika, turunan suatu fungsi

$$ f(x) $$

didefinisikan sebagai lanju perubahan fungsi terhadap $x$:

$$ \frac{df}{dx} = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h} $$

Secara sederhana:

> **Turunan mengukur seberapa cepat suatu besaran berubah terhadap besaran lainnya.**

Dalam fisika, konsep ini sangat fundamental.

Misalnya posisi benda:

$$ x = x(t) $$

maka kecepatan adalah turunan posisi terhadap waktu:

$$ v(t) = \frac{dx}{dt} $$

dan percepatan adalah:

$$ \alpha(t) = \frac{dv}{dt} = \frac{d^2 x}{dt^2} $$

Jadi, jika kita mempunyai data posisi benda terhadap waktu, kita dapat menggunakan **turunan numerik untuk menghitung kecepatan dan percepatan**

# **2. Mengapa kita Membutuhkan Turunan Numeri?**

Dalam kalkulus, kita biasanya mempunyai fungsi analitik.

Contoh:

$$ f(x) = x^2 $$

Kita dapat langusng menghitung:

$$ f'(x) = 2x $$

Tetapi dalam fisika komputasi, sering kali kita tidak mempunyai fungsi analitik.

Misalnya hasil eksperimen:

<div align="center">

| $t (s)$ | $x (m)$ |
| :---: | :---: |
| 0.0 | 0.00 |
| 0.1 | 0.05 |
| 0.2 | 0.20 |
| 0.3 | 0.45 |
| 0.4 | 0.80 |

</div>

Kita hanya mempunyai **data diskrit**.

Kita tidak mempunyai persamaan eksplisit:

$$ x = x(t) $$

Tetapi kita ingin mengetahui:

$$ v = \frac{dx}{dt} $$

Di sinilah **turunan numerik** digunakan

#### **Prinsip Utama**

Turunan numerik mencoba mendekati:

$$ \boxed{\frac{df}{dt}} $$

menggunakan nilai fungsi yang diketahui pada beberapa titik.

# **3. Forward Difference**

Metode pertama yang kita pelajari adalah **forward difference**

Dari definisi turunan:

$$ f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h} $$

Jika $h$ dibuat cukup kecil, kita dapat menggunakan pendekatan:

$$ \boxed{f'(x) = \frac{f(x+h) - f(x)}{h}} $$

Ini disebut **forward difference** karena kita menggunakan titik di depan $x$.

**Contoh**  
Misalkan:

$$ f(x) = x^2 $$

dan lita ingin menghitung turunan di:

$$ x = 2 $$

Gunakann:

$$ h = 0.01 $$

Maka:

$$ f'(2) \approx \frac{f(2.01) - f(2)}{0.01} $$

Karena:
 * $ f(2.01) = 2.01^2 = 4.0401 $
 * $ f(2) = 2^2 = 4 $

maka:

$$ f'(2) \approx \frac{4.0401 - 4}{0.01} = 4.01 $$

Sedangkan hasil analitik:

$$ f'(x) = 2x $$

sehingga:

$$ f'(2) = 4 $$

Jadi terdapat error:

$$ 4.01 - 4 = 0.01 $$

# **4. Backward Difference**

Metode kedua adalah **backward difference**

Kita menggunakan titik sebelum $x$:

$$ x-h $$

Formula:

$$ \boxed{f'(x) \approx \frac{f(x) - f(x-h)}{h}} $$

# **5. Central Difference**

Daripada hanya menggunakan titik di depan atau di belakang, kita menggunakan **kedua titi**:

$$ x-h $$

dan

$$ x+h $$

Formula:

$$ \boxed{f'(x) \approx \frac{f(x+h) - f(x-h)}{2h}} $$

Metode ini disebut **central difference**

#### **Mengapa Central Difference Lebih AKurat?**

Ini dapat ditunjukkan menggunakan **Taylor series**

Untuk $f(x+h)$:

$$f(x+h) = f(x) + hf'(x) + \frac{h^2}{2}f''(x) + \frac{h^3}{6}f'''(x) + \dots$$

Sedangkan:

$$f(x-h) = f(x) - hf'(x) + \frac{h^2}{2}f''(x) - \frac{h^3}{6}f'''(x) + \dots$$

Kurangkan kedua persamaan:

$$f(x+h) - f(x-h) = 2hf'(x) + \frac{h^3}{3}f'''(x) + \dots$$

Maka:

$$\boxed{f'(x) = \frac{f(x+h) - f(x-h)}{2h} - \frac{h^2}{6}f'''(x) + \dots}$$

Sehingga error utamanya adalah:

$$\boxed{O(h^2)}$$

# **Implementasi Python**

```python
def f(x):
    return x**2

def df_analitik(x):
    return 2*x
```

```python
def forward_difference(f,x,h):
    return (f(x + h) - f(x)) / h

x = 2.0
h = 0.01

hasil = forward_difference(f,x,h)

print(f"Turunan numerik  = {hasil}")
print(f"Turunan analitik = {df_analitik(x)}")
```

```python
def backward_difference(f,x,h):
    return (f(x) - f(x - h)) / h

x = 2.0
h = 0.01

hasil = backward_difference(f, x, h)

print(f"Turunan numerik  = {hasil}")
print(f"Turunan analitik = {df_analitik(x)}")
```

```python
def central_difference(f, x, h):
    return (f(x + h) - f(x - h)) / (2*h)

x = 2.0
h = 0.01

hasil = central_difference(f, x, h)

print(f"Turunan numerik  = {hasil}")
print(f"Turunan analitik = {df_analitik(x)}")
```

```python
x = 2.0
h = 0.01

analitik = df_analitik(x)

forward = forward_difference(f, x, h)
backward = backward_difference(f, x, h)
central = central_difference(f, x, h)

print(f"Hasil analitik : {analitik}\n")

print("Forward difference")
print(f"Hasil = {forward}")
print(f"Error = {abs(forward - analitik)}\n")

print("Backward difference")
print(f"Hasil = {backward}")
print(f"Error = {abs(backward - analitik)}\n")

print("Central difference")
print(f"Hasil = {central}")
print(f"Error = {abs(central - analitik)}")
```

### **Turunan Numerik pada Data Eksperimen**

Sekarang kita masuk ke aplikasi yang jauh lebih dekat dengan fisika komputasi

Misalkan kita mempunyai data:

```python
import numpy as np

t = np.array([
    0.0,
    0.1,
    0.2,
    0.3,
    0.4
])

x = np.array([
    0.00,
    0.05,
    0.20,
    0.45,
    0.80
])
```

Untuk titik interior:

$$ \boxed{v_i \approx \frac{x_{i+1} - x_{i-1}}{2\Delta t}} $$

```python
dt = 0.1

v = np.zeros(len(x))

# Forward difference untuk titik pertama
v[0] = (x[1] - x[0]) / dt

# Central difference untuk titik interior
for i in range(1, len(x) - 1):
    v[i] = (x[i+1] - x[i-1]) / (2*dt)

# Backward difference untuk titik terakhir
v[-1] = (x[-1] - x[-2]) / dt

print(v)
```

Dalam praktiknya, Python sudah menyediakan fungsi untuk menghitung gradien numerik:

```python
import numpy as np

v = np.gradient(x, t)

print(v)
```

### **Turunan Kedua Secara Numerik**

Kita juga dapat menghitung turunan kedua.

Formula central difference:

$$ \boxed{f''(x) \approx \frac{f(x+h) - 2f(x) + f(x+h)}{h^2}} $$

Misalnya:

$$ f(x) = x^2 $$

maka secara analitik:

$$ f''(x) = 2 $$

Implementasi:

```python
def second_derivative(f,x,h):
    return (f(x + h) - 2*f(x) + f(x - h)) / h**2

x = 2
h = 0.01

hasil = second_derivative(f,x,h)

print(f"Turunan kedua = {hasil}")
```
